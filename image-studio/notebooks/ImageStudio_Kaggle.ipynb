# Image Studio — Kaggle GPU

One-click prompt-to-image using **Z-Image Turbo** (real image generation, not the CPU test model). Choose a GPU runtime, enable internet, run cells in order. Outputs are saved in the session output folder. 

```python
import os, subprocess, sys
from pathlib import Path
if not Path('/content').exists():
    raise RuntimeError('This notebook is designed for Google Colab. Open it in Colab.')
REPO = Path('/content/universal-ai-game-studio')
BRANCH = 'image-generation-studio'
URL = 'https://github.com/dvd2pe/universal-ai-game-studio.git'
if REPO.exists():
    subprocess.run(['git','-C',str(REPO),'fetch','origin',BRANCH],check=True)
    subprocess.run(['git','-C',str(REPO),'checkout',BRANCH],check=True)
    subprocess.run(['git','-C',str(REPO),'pull','--ff-only','origin',BRANCH],check=True)
else:
    subprocess.run(['git','clone','--depth','1','--branch',BRANCH,URL,str(REPO)],check=True)
OUTPUT_DIR='/content/outputs'
Path(OUTPUT_DIR).mkdir(parents=True,exist_ok=True)
os.environ['IMAGE_STUDIO_OUTPUT']=OUTPUT_DIR
print('Repo ready:',REPO)
print('Output folder:',OUTPUT_DIR)
```

## Install dependencies

The notebook installs the app's pinned/declared requirements. Initial installation and model download can take a while.

```python
subprocess.check_call([sys.executable,'-m','pip','install','-q','-r',str(REPO/'image-studio'/'requirements.txt')])
print('Dependencies installed. Restart runtime only if Colab explicitly asks; then rerun setup cells.')
```

## Launch Image Studio

Pick **Z-Image Turbo** in the UI for a first photorealistic prompt. Use the public Gradio link printed by the app. Keep this cell running.

```python
os.chdir(REPO/'image-studio')
subprocess.run([sys.executable,'app.py'],check=True)
```
