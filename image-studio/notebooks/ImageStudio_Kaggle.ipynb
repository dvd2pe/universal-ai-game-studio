# Open Image Studio — Kaggle launcher
Enable Internet and a GPU accelerator, then run the single cell below. Supports Z-Image Turbo, Qwen Image 2.1 Turbo and Qwen Image Edit 2509. No ComfyUI.

```python
import os, subprocess, sys
from pathlib import Path
output_dir = '/kaggle/working/outputs'
Path(output_dir).mkdir(parents=True, exist_ok=True)
os.environ['IMAGE_STUDIO_OUTPUT'] = output_dir
repo = 'https://github.com/dvd2pe/universal-ai-game-studio.git'
branch = 'image-generation-studio'
work = Path('/kaggle/working/universal-ai-game-studio')
if work.exists():
    subprocess.run(['git','-C',str(work),'fetch','origin',branch],check=True)
    subprocess.run(['git','-C',str(work),'checkout',branch],check=True)
    subprocess.run(['git','-C',str(work),'pull','--ff-only','origin',branch],check=True)
else:
    subprocess.run(['git','clone','--depth','1','--branch',branch,repo,str(work)],check=True)
subprocess.check_call([sys.executable,'-m','pip','install','-q','-r',str(work/'image-studio'/'requirements.txt')])
os.chdir(work/'image-studio')
print('Outputs:',output_dir)
print('Starting UI. Keep this cell running while you use it.')
subprocess.run([sys.executable,'app.py'],check=True)
```
